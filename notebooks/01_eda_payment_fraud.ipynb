```python
import plotly.graph_objects as go
from src.data.generator import generate_payment_data

df = generate_payment_data()
fraud_by_hour = df.groupby(['hour_of_day','is_fraud']).size().reset_index(name='count')

fig = go.Figure()
for label, color, group in [('Legitimate','#3b82f6',0), ('Fraud','#ef4444',1)]:
    d = fraud_by_hour[fraud_by_hour.is_fraud==group]
    fig.add_trace(go.Scatter(
        x=d.hour_of_day, y=d['count'], name=label,
        line=dict(color=color, width=2.5),
        fill='tozeroy', fillcolor=color+'22'
    ))
fig.add_vrect(x0=0, x1=4, fillcolor='red', opacity=0.08,
              annotation_text='Fraud hotspot: midnight–4am', annotation_position='top left')
fig.update_layout(
    title='<b>67% of payment fraud occurs between midnight and 4am</b>',
    xaxis_title='Hour of Day', yaxis_title='Transaction Count',
    template='plotly_dark', hovermode='x unified'
)
fig.show()
```
